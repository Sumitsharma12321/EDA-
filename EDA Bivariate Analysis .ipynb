## Bivariate analysis

```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
train = pd.read_csv()
```

```python
flights = sns.load_dataset('flights')
```

```python
iris = sns.load_dataset('iris')
```

## Scatter plot(Numerical - Numerical)

```python
sns.scatterplot(x = 'total_bill', y = 'tip' , data = tips)
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex', data = tips)
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex',style = 'smoker', size = 'size' , data = tips )
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
titanic.dtypes
```

```python
sns.barplot(x = 'pclass', y = 'age',data = titanic, palette = 'Set2')
```

## 3. Box Plot(N - C)

```python
sns.boxplot(x='sex',y = 'age', data = titanic ,hue = 'survived', palette = 'Set2')
```

## Distplot (N_C)

```python
sns.distplot(titanic[titanic['survived']==0]['age'],hist = False)
sns.distplot(titanic[titanic['survived']==1]['age'],hist = False)
```

## 5.Heatmap (C-C)

```python
sns.heatmap(pd.crosstab(titanic['pclass'],titanic['survived']))
```

```python
(titanic.groupby('pclass')['survived'].mean() * 100).plot(kind = 'bar')
```

```python
(titanic.groupby('sex')['survived'].mean() * 100)
```

```python
(titanic.groupby('embarked')['survived'].mean() * 100)
```

## ClusterMap (C-C)

```python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['survived']))
```

## pairplot

## Line plot (N-N) - for time based 

```python
flights.head()
```

```python
new = flights.groupby('year').sum(numeric_only = True).reset_index()
```

```python
sns.lineplot(x= 'year', y = 'passengers', data = new)
```

## Using Pivot Table 

```python
sns.clustermap(flights.pivot_table(values='passengers', index='month',columns='year'))
```
