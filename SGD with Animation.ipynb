```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression

import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
from matplotlib.animation import FuncAnimation
import matplotlib.animation as animation
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

```python

plt.scatter(X,y)
plt.show()
```

```python
lr = LinearRegression()
lr.fit(X,y)
print(lr.coef_)
print(lr.intercept_)
```

```python
import time

b = 150
m = -127.82
all_b = []
all_m = []
all_cost = []
all_lr = []

epochs = 1

start = time.time()

t0,t1 = 5,50
def learning_rate(t):
    return t0/(t + t1)

for i in range(epochs):
    for j in range(X.shape[0]):
        
        lr = learning_rate(i * X.shape[0] + j)
        
        idx = np.random.randint(X.shape[0], size=1)
        
        slope_b = -2 * (y[idx] - (m * X[idx]) -  b)
        slope_m = -2 * (y[idx] - (m * X[idx]) -  b) * X[idx]
        cost = (y[idx] - m * X[idx] -b) ** 2
        
        
        b = b - (lr * slope_b)
        m = m - (lr * slope_m)
        all_b.append(b)
        all_m.append(m)
        all_cost.append(cost)
        all_lr.append(lr)
print("Total time taken",time.time() - start)
```

```python
from IPython.display import HTML

fig, ax = plt.subplots(figsize=(9,5))
x_i = np.arange(-3, 3, 0.1)
y_i = x_i*(-27) -150
ax.scatter(X, y)
line, = ax.plot(x_i,x_i * 50 - 4,'r-',linewidth=2)

def update(i):

    label = f'Epoch {i+1}'
    line.set_ydata(x_i * all_m[i] + all_b[i])
    ax.set_xlabel(label)
    return line,

anim = FuncAnimation(
    fig,
    update,
    frames=len(all_m),   # 100 hardcode karne ki zarurat nahi
    interval=50,
    blit=True,
    repeat=False
)

HTML(anim.to_jshtml())
```

```python
m_arr = np.linspace(-150, 150, 10)
b_arr = np.linspace(-150, 150, 10)
mGrid, bGrid = np.meshgrid(m_arr,b_arr)

final = np.vstack((mGrid.ravel().reshape(1,100),bGrid.ravel().reshape(1,100))).T

z_arr = []

for i in range(final.shape[0]):
    z_arr.append(np.sum((y - final[i,0]*X.reshape(100) - final[i,1])**2))

z_arr = np.array(z_arr).reshape(10,10)
```

```python
%matplotlib inline

from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib.pyplot as plt

# Contour Graph
fig, axis = plt.subplots(figsize=(9,5))

axis.set_xlim(-150, 150)
axis.set_ylim(-150, 150)

axis.contourf(m_arr, b_arr, z_arr)

line, = axis.plot([], [], lw=2, color='white')

xdata, ydata = [], []

def animate(i):

    xdata.append(all_m[i])
    ydata.append(all_b[i])

    line.set_data(xdata, ydata)

    axis.set_title(f"Epoch {i+1}")

    return line,

anim = FuncAnimation(
    fig,
    animate,
    frames=len(all_m),   # hardcoded 100 ki jagah
    interval=100,
    repeat=False,
    blit=True
)

HTML(anim.to_jshtml())
```
